```python
import math
```

```python
# -----------------------------------
# Test for Single Proportion
# -----------------------------------
n = 400
x = 220
p0 = 0.50
phat = x / n
z = (phat - p0) / math.sqrt((p0 * (1 - p0)) / n)
print("Single Proportion Test")
print("Sample Proportion =", round(phat, 3))
print("Z statistic =", round(z, 3))
# -----------------------------------
# Difference of Proportions Test
# -----------------------------------
n1 = 500
x1 = 280
n2 = 400
x2 = 200
p1 = x1 / n1
p2 = x2 / n2
p = (x1 + x2) / (n1 + n2)
z_diff = (p1 - p2) / math.sqrt(
 p * (1 - p) * ((1 / n1) + (1 / n2))
)
print("\nDifference of Proportions Test")
print("Pooled Proportion =", round(p, 3))
print("Z statistic =", round(z_diff, 3))
```
