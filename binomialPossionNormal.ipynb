```python
# Program for probability analysis of Binomial,

# Poisson, and Normal distributions

from scipy.stats import binom, poisson, norm

# -------------------------------

# Binomial Distribution

# -------------------------------

n = 12      # number of trials
p = 0.6     # probability of success
x = 7       # number of successes
binomial_prob = binom.pmf(x, n, p)
print("BINOMIAL DISTRIBUTION")
print("----------------------")
print("Probability P(X=7) =", binomial_prob)
print("Mean =", n * p)
print("Variance =", n * p * (1 - p))

# -------------------------------

# Poisson Distribution

# -------------------------------

lam = 5     # average rate
x = 3
poisson_prob = poisson.pmf(x, lam)
print("\nPOISSON DISTRIBUTION")
print("----------------------")
print("Probability P(X=3) =", poisson_prob)
print("Mean =", lam)
print("Variance =", lam)

# -------------------------------

# Normal Distribution

# -------------------------------

mu = 70
sigma = 8
x = 75
normal_prob = norm.pdf(x, mu, sigma)
print("\nNORMAL DISTRIBUTION")
print("----------------------")
print("Probability Density at x=75 =", normal_prob)
print("Mean =", mu)
print("Variance =", sigma ** 2)
```
