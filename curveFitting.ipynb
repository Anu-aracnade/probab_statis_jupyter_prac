```python
import numpy as np
import matplotlib.pyplot as plt

# Input data
x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([1.8, 4.7, 8.5, 12.6, 18.3, 24.1])

# Second-degree parabola fitting
coeff_parabola = np.polyfit(x, y, 2)

# General curve fitting (cubic)
coeff_general = np.polyfit(x, y, 3)

# Smooth x-values
x_fit = np.linspace(min(x), max(x), 100)

# Predicted y-values
y_parabola = np.polyval(coeff_parabola, x_fit)
y_general = np.polyval(coeff_general, x_fit)

# Print equations
print("Second-Degree Parabola Coefficients:")
print(coeff_parabola)

print("\nGeneral Curve Coefficients:")
print(coeff_general)

# Plotting
plt.figure(figsize=(8, 5))

plt.scatter(x, y, color='blue', marker='o', s=55, label='Actual Data')
plt.plot(x_fit, y_parabola, label='Parabolic Fit', linewidth=2)
plt.plot(x_fit, y_general, '--', label='Cubic Curve Fit', linewidth=2)

plt.xlabel('Independent Variable (x)')
plt.ylabel('Dependent Variable (y)')
plt.title('Curve Fitting using Least Squares')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)

plt.show()
```
